````python
# Q1> print your name x no's of times

name = input("give me ur name")

print(name)
````

````python
name = input("give me ur name")
num = int(input("no's of times you want to print"))

for i in range(num):
    print(i)
  
    print(name)

# i -> variable  start from 0 to n-1
````

````python
# Start and stop

name = input("give me ur name")
num = int(input("no's of times you want to print"))

for i in range(3,num):
    print(i)
  
    print(name)

# range(p1, p2)
# p1-> start value and p2 -> end value
````

````python
# Step

# range(p1, p2, p3)
# p1-> start value, p2 -> end value and p3 -> step value

name = input("give me ur name")
num = int(input("no's of times you want to print"))

for i in range(1,num, 3):
    print(i)
  
    print(name)
````

````python
# reverse

# range(p1, p2, -1)
# p1-> start value last end 10 , p2 ->  where to stop 0 and p3 -> step -ve value

# for variable in itrable():
#   your logic

name = input("give me ur name")
num = int(input("no's of times you want to print"))

#(0,10,1)
#(10,2,-1) -> reverse order
for i in range(num,0,-1):
    print(i)
  
    print(name)
````

````python
# we can run for loops with data types like list, strings, dict, tupple, set

dog = "bob the dog"
print(dog[10])
print(len(dog))
print("-----")

for i in dog:
    print(i)
````
