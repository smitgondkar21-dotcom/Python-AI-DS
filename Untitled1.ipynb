```python
# Largest of Three Numbers (without max, TCS, Wipro, Infosys)

a = int(input("first number"))
b = int(input("2nd number"))
c = int(input("3rd number"))

if a>b and a>c:
    print("1st no is the largest")
elif b>a and b>c:
    print("2nd no is largest")
else:
    print("3rd no is largest")
```

```python
# Second largest of 3 without sorting (TCS, Infosys)

a = int(input("first number"))
b = int(input("2nd number"))
c = int(input("3rd number"))

if a>b and a>c:
    if b>c:
        print("b is 2nd largest no")
    else:
        print("c is the 2nd largest")

elif b>a and b>c:
    if a>c:
        print("2nd largest a")
    else:
        print("2nd largest is c")
elif c>a and c>b:
    if a>b:
        print("a is 2nd largest")
    else:
        print("b is 2nd largest")
else:
    print("two number or three numbers are equal")
```
